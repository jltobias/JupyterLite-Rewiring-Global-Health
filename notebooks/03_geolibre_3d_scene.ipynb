## 3D scene as a stack of contracts

A global-health scene can combine terrain, buildings, population, mobility, facilities, climate/environment, and model outputs. Each layer should carry provenance, time, units, spatial resolution, and a privacy classification.

```python
scene=pd.DataFrame({
"layer":["terrain","buildings","population","roads/transit","facilities","health model output"],
"formats":["COG/terrain tiles","GeoJSON/PMTiles","Zarr/COG/H3","PMTiles/GeoJSON","GeoJSON","COG/Zarr/GeoJSON"],
"privacy":["low","low","medium","low","medium","high if precise"]
})
scene
```
